```python
# Loading the dataset using pandas
import pandas as pd
from pathlib import Path

# Works whether Jupyter is launched from the repository root or notebooks/.
data_path = Path("data/raw/customer_shopping_behavior.csv")
if not data_path.exists():
    data_path = Path("../data/raw/customer_shopping_behavior.csv")

df = pd.read_csv(data_path)
```

```python
df['Review Rating']=df.groupby("Category")['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isna().sum()
```

```python
# Make columns more consistent and readable for avoiding any future errors or problems
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_")
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# creating a new column for age group
bins = [17, 25, 45, 60, 100]
age_category = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']

# Split the data
df['age_group'] = pd.cut(df['age'], bins=bins, labels=age_category)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create a column for purchase_frequency_days
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

```python
df[['discount_applied','promo_code_used']].value_counts()
```

```python
# Above analysis clearly shows there are redundant columns so we drop one column
df = df.drop(columns='promo_code_used')
```

```python
import os
import pandas as pd
from sqlalchemy import URL, create_engine, text

# 1. MySQL connection details
username = os.getenv("MYSQL_USER", "root")
password = os.getenv("MYSQL_PASSWORD", "")  # Set this in your environment; do not commit passwords
host = os.getenv("MYSQL_HOST", "localhost")
port = int(os.getenv("MYSQL_PORT", "3306"))
database = os.getenv("MYSQL_DATABASE", "customer_behavior_analysis")

# 2. Create SQLAlchemy engine
connection_url = URL.create(
    drivername="mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

engine = create_engine(connection_url)

# 3. Test MySQL connection
with engine.connect() as connection:
    result = connection.execute(text("SELECT 1"))
    print("MySQL connection successful:", result.scalar())

# 4. Write cleaned DataFrame to MySQL
table_name = "customer_data"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",
    index=False
)

print("Data successfully imported into MySQL!")

# 5. Read data back from MySQL
sql_query = "SELECT * FROM customer_data LIMIT 5;"

customer_data = pd.read_sql(sql_query, engine)

print(customer_data)

# 6. Close engine connections
engine.dispose()
```
